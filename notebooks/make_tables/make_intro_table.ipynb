# Format the LaTeX table introducing the sample

This notebook has a very simple and limited purpose, to construct a correctly formatted table of the sources in the LoVoCCS sample, with basic properties - this essentially is what we use in papers to mostly represent the knowledge we had going in to the analyses (i.e. few newly measured properties).

## Import Statements

```python
import pandas as pd
import numpy as np
import os
```

## Checking that another X-LoVoCCS repository is present

This notebook relies on the LoVoCCS observation summaries created in the X-LoVoCCS-Data repository (which contains all the notebooks necessary to recreate the LoVoCCS X-ray dataset) - thus we need to ensure that the repository is where we expect it to be before this notebook continues (this is mostly in case anyone ever wishes to recreate all of our analyses):

```python
if not os.path.exists("../../../X-LoVoCCS-Data/"):
    raise FileNotFoundError("We require the X-LoVoCCS-Data repository to be cloned into the same "\
                            "parent folder as the X-LoVoCCS-Analyses directory.")
else:
    cov_summ_path = "../../../X-LoVoCCS-Data/outputs/summary_files/"
```

## Loading data files

Important considerations for this dataset:

* We only need the base X-LoVoCCS sample file and the X-ray mission coverage file from this repository.
* However we also need some observation archive summary files from the X-LoVoCCS-Data repository.

### X-LoVoCCS-I base sample

```python
xlovoccs_base = pd.read_csv("../../sample_files/X-LoVoCCSI.csv")
xlovoccs_base.insert(0, 'name', xlovoccs_base['LoVoCCSID'].apply(lambda x: "LoVoCCS-" + str(x)))
xlovoccs_base.head(6)
```

### X-LoVoCCS coverage measurements

```python
xlovoccs_xcov_frac = pd.read_csv("../../outputs/result_files/misc/mission_coverage_fractions.csv")
xlovoccs_xcov_frac.head(6)
```

### X-LoVoCCS numbers of X-ray observations

```python
xlovoccs_num_obs = pd.read_csv(cov_summ_path + "each_cluster_obs_num.csv")
xlovoccs_num_obs['parent_LoVoCCSID'] = xlovoccs_num_obs['name'].apply(lambda x: int(x.split('-')[-1]))
del xlovoccs_num_obs['name']
xlovoccs_num_obs.head(6)
```

### X-LoVoCCS total exposure times

```python
xlovoccs_exp_times = pd.read_csv(cov_summ_path + "total_xmm_exposures.csv")
xlovoccs_exp_times.head(6)
```

### Combining tables

```python
xlovoccs_samp = pd.merge(xlovoccs_base, xlovoccs_xcov_frac, left_on='name', right_on='name', how='outer')
xlovoccs_samp = pd.merge(xlovoccs_samp, xlovoccs_num_obs, left_on='parent_LoVoCCSID', 
                         right_on='parent_LoVoCCSID', how='outer')
xlovoccs_samp = pd.merge(xlovoccs_samp, xlovoccs_exp_times, left_on='name', right_on='name', how='outer')
xlovoccs_samp['exp'] = xlovoccs_samp['exp'].fillna('0.0 ks')

# Makes sure that the order is correct
xlovoccs_samp = xlovoccs_samp.set_index('name').loc[xlovoccs_base['name'].values].reset_index(drop=False)
xlovoccs_samp.head(6)
```

## Generating the table string

Here we generate the string of LaTeX code that can be copied in as the data body of the table:

```python
round_pos = 3
```

```python
lines = ""
for row_ind, row in xlovoccs_samp.iterrows():
    rel_name = row['Name']
    # Shorten the name
    if rel_name == "RX J0820.9+0751/RXC J0821.0+0751":
        rel_name = "RXC J0821.0+0751"
    
    # Add a dagger to match the caption - tells the reader that this one has another name
    if rel_name == 'RXC J2218.2-0350 East':
        rel_name = "RXC J2218.2-0350 East$^{\dagger}$"
    
    # Adds spaces around north/south/east/west/centre name components
    comps = {'North': r"\textit{North}", 'South': r"\textit{South}", 'East': r"\textit{East}", 
             'West': r"\textit{West}", 'Centre': r"\textit{Centre}"}
    
    # Never said I would make this auto table making beautiful...
    for comp, comp_rep in comps.items():
        # Adds spaces before the north/south/etc. component
        if comp in rel_name and " " + comp not in rel_name:
            rel_name = rel_name.replace(comp, ' '+comp_rep)
        elif comp in rel_name:
            rel_name = rel_name.replace(comp, comp_rep)

    rel_lov_id = row['LoVoCCSID']
    
    res_line = "{n} & {i} & ".format(n=rel_name, i=rel_lov_id)
    
    ra = round(row['start_ra'], round_pos)
    dec = round(row['start_dec'], round_pos)
    res_line += "{r} & {d} & ".format(r=ra, d=dec)
    
    res_line += "{} & ".format(round(row['MCXC_Redshift'], 3))
#     if not rel_lov_id[-1].isdigit() and rel_lov_id[-1] != 'A':
#         res_line += "- & "
#     else:
#         res_line += "{} & ".format(round(row['MCXC_R500'], 2))
    
    
    # For some clusters we are bolding their coverage fraction and adding a ddagger
    if row['xmm_pointed_1.0Mpc'] < 0.4:
        cov_str = r"\textbf{{{v}}}$^{{\ddagger}}$".format(v=row['xmm_pointed_1.0Mpc'])
    else:
        cov_str = row['xmm_pointed_1.0Mpc']
    
    res_line += "{tc} & {nxmm} & {texp} ".format(tc=cov_str, nxmm=row['xmm_pointed'], 
                                                 texp=round(float(row['exp'].split(' ')[0]), 1))
        
    res_line += '\\\ \n% \n'
    lines += res_line
print(lines)
```
